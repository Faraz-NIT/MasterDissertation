# Evidence-gated replenishment walkthrough
Run from the repository root or its notebooks directory. This uses synthetic data and no LLM. It is not dissertation evidence.

```python
from pathlib import Path
import os, sys
root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
os.chdir(root)
sys.path.insert(0, str(root / "src"))
from ega.config import ExperimentConfig
from ega.data.synthetic import make_demo
from ega.experiment import run_experiment
from ega.evaluation.report import render_report
from ega.evaluation.faithfulness import replay, counterfactual
```

```python
make_demo("data/processed/notebook", items=1, stores=2, days=180)
config = ExperimentConfig.model_validate({"dataset":"data/processed/notebook", "output":"results/notebook",
    "policies":["D0","D1"], "scenarios":["normal","derived_field_collapse"], "seeds":[7],
    "days":4, "warmup_days":2, "solver":{"horizon":6,"scenarios":3}})
summary = run_experiment(config)
summary[["policy","scenario","cost","fill_rate","held_decisions","harmful_executions"]]
```

```python
render_report(config.output)
replay("results/notebook/D1__normal__seed7__origin0", day=140)
```

```python
counterfactual("results/notebook/D1__normal__seed7__origin0", day=140, factor="budget", multiplier=0.5)
```
